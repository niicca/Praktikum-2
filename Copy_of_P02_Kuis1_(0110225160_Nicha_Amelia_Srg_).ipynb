---
### Tugas 1 (15 poin) — Memuat dan memeriksa

```python
# a. Cetak jumlah baris dan kolom
print("Jumlah baris dan kolom:", df.shape)

# b. Cetak jumlah nilai kosong (missing) untuk setiap kolom
print(df.isna().sum())

# c. Cetak berapa mahasiswa unik (nim_hash) yang pernah berkunjung
print(df['nim_hash'].nunique())
```

---
### Tugas 2 (20 poin) — Filter

```python
# Ambil kunjungan mahasiswa prodi 'Teknik Informatika' dengan durasi_menit > 120.
# Cetak jumlah barisnya dan rata-rata jumlah_pinjam pada kelompok tersebut (2 desimal).

hasil = df[
    (df['prodi'] == 'Teknik Informatika') &
    (df['durasi_menit'] > 120)]

print("Jumlah baris:", len(hasil))
print("Rata-rata jumlah_pinjam:", round(hasil['jumlah_pinjam'].mean(), 2))
```

---
### Tugas 3 (20 poin) — Kolom turunan

```python
# a. Buat kolom 'durasi_jam' = durasi_menit / 60, dibulatkan 2 desimal
df['durasi_jam'] = (df['durasi_menit'] / 60).round(2)
print(df['durasi_jam'])

# b. Buat kolom 'lama' berisi 'Panjang' jika durasi_menit >= 90, selain itu 'Singkat'
df['lama'] = np.where(
df['durasi_menit'] >= 90,'Panjang','Singkat')
print(df['lama'])


# c. Tampilkan 5 kunjungan dengan durasi terlama (kolom: prodi, jenis_ruang, durasi_jam)
print(df.sort_values('durasi_jam', ascending=False)[['prodi', 'jenis_ruang', 'durasi_jam']].head(5))
```

---
### Tugas 4 (25 poin) — Agregasi

```python
# Buat tabel ringkasan per 'jenis_ruang' yang memuat:
#   - jumlah kunjungan
#   - rata-rata durasi_menit (1 desimal)
#   - total jumlah_pinjam
#   - rata-rata kepuasan (2 desimal)
# Urutkan dari rata-rata durasi terbesar. Simpan ke variabel 'ringkas' dan tampilkan.


ringkas = df.groupby('jenis_ruang').agg(
                    jumlah_kunjungan=('jenis_ruang', 'count'),
                    rata_durasi=('durasi_menit', 'mean'),
                    total_pinjam=('jumlah_pinjam', 'sum'),
                    rata_kepuasan=('kepuasan', 'mean')).round({'rata_durasi': 1,'rata_kepuasan': 2}
                    ).sort_values('rata_durasi', ascending=False)
print(ringkas)
```

---
### Tugas 5 (10 poin) — Menyimpan hasil

```python
# Simpan tabel 'ringkas' ke file 'ringkasan_ruang.csv'
ringkas.to_csv('ringkasan_ruang.csv', index=False)
print("File berhasil disimpan sebagai ringkasan_ruang.csv")
```
